```python
from pathlib import Path
import pandas as pd
import numpy as np

# Get the project root directory
PROJECT_ROOT = Path.cwd()

# If Jupyter is running from the notebooks folder, go one level up
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

# Define data directories
RAW_DATA_PATH = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA_PATH = PROJECT_ROOT / "data" / "processed"

# Create processed directory if it doesn't exist
PROCESSED_DATA_PATH.mkdir(parents=True, exist_ok=True)

# Files to load
files = {
    "sales": "Sales.csv",
    "customers": "Customers.csv",
    "products": "Products.csv",
    "stores": "Stores.csv",
    "exchange_rates": "Exchange_Rates.csv"
}

data = {}

for name, filename in files.items():

    # Build the complete file path
    file_path = RAW_DATA_PATH / filename

    # Check whether the file exists
    if not file_path.exists():
        raise FileNotFoundError(f"Missing file: {file_path}")

    # Load the CSV file
    data[name] = pd.read_csv(file_path, encoding="cp1252")
    print(f"{name}: loaded successfully")

print("\nAll datasets loaded successfully!")
```

```python
def data_quality_report(df):

    report = pd.DataFrame({
        "Data Type": df.dtypes.astype(str),
        "Missing Count": df.isnull().sum(),
        "Missing Percentage": (
            df.isnull().mean() * 100
        ).round(2),
        "Unique Values": df.nunique(),
        "Duplicate Count": [
            df[col].duplicated().sum()
            for col in df.columns
        ]
    })

    return report
```

```python
for name, df in data.items():

    print(f"\nDATA QUALITY REPORT: {name.upper()}")

    display(data_quality_report(df))
    
```

## Relational Integrity Checks

```python
primary_keys = {
    "customers": "CustomerKey",
    "products": "ProductKey",
    "stores": "StoreKey"
}

for table_name, key in primary_keys.items():

    df = data[table_name]

    print(f"\n{table_name.upper()}")
    print(f"Key: {key}")
    print(f"Unique: {df[key].is_unique}")
    print(f"Missing: {df[key].isnull().sum()}")
    print(f"Duplicate keys: {df[key].duplicated().sum()}")
```

## Validate the Sales foreign keys

```python
relationships = {
    "CustomerKey": ("customers", "CustomerKey"),
    "ProductKey": ("products", "ProductKey"),
    "StoreKey": ("stores", "StoreKey")
}

sales = data["sales"]

for sales_column, (table_name, dimension_column) in relationships.items():

    valid_keys = set(data[table_name][dimension_column])

    invalid_rows = ~sales[sales_column].isin(valid_keys)

    print(f"\nRelationship: Sales.{sales_column}")
    print(f"Invalid references: {invalid_rows.sum()}")
    print(f"Percentage: {invalid_rows.mean() * 100:.2f}%")
    print(f"Invalid values: {sales.loc[invalid_rows, sales_column].unique()[:10]}")
```

# Investigate the missing delivery dates
Let's examine whether missing delivery dates are associated with particular stores or transaction patterns.

```python
sales = data["sales"]

sales["Delivery Missing"] = sales["Delivery Date"].isna()

print(
    sales.groupby("Delivery Missing")["StoreKey"]
    .agg(["count", "nunique"])
)
```

```python
display(
    sales[sales["Delivery Missing"]]
    .groupby("StoreKey")
    .size()
    .sort_values(ascending=False)
    .head(15)
)
```

```python
store_delivery_report = (
    sales.groupby("StoreKey")
    .agg(
        Total_Sales=("Delivery Date", "size"),
        Missing_Delivery=("Delivery Date", lambda x: x.isna().sum())
    )
)

store_delivery_report["Missing_Percentage"] = (
    store_delivery_report["Missing_Delivery"] /
    store_delivery_report["Total_Sales"] * 100
).round(2)

display(
    store_delivery_report.sort_values(
        "Missing_Percentage",
        ascending=False
    )
)
```
